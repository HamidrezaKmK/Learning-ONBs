```python
# Cell 2 — load CelebA-HQ (streaming)
from datasets import load_dataset

# A commonly available HF mirror of CelebA-HQ (30k images)
ds = load_dataset("mattymchen/celeba-hq", split="train", streaming=True)
it = iter(ds)

sample = next(it)
sample.keys(), type(sample["image"])
```

```python
# Cell 3 — visualize a few samples
import matplotlib.pyplot as plt

n = 6
imgs = []
for _ in range(n):
    imgs.append(next(it)["image"])

plt.figure(figsize=(12, 4))
for i, im in enumerate(imgs):
    plt.subplot(1, n, i+1)
    plt.imshow(im)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Cell 4 — check resolution of one image
im = imgs[0]
print("Resolution:", im.size)  # (W, H)
```
